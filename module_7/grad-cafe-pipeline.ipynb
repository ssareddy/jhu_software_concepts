# Grad Café S3 → SageMaker Pipeline

This notebook downloads the applicant dataset from S3 using boto3
and saves it locally as `applicant_data_SM.json`.

## Setup
Credentials are loaded from environment variables — never hardcoded.
Set the following before running:
- `AWS_ACCESS_KEY_ID`
- `AWS_SECRET_ACCESS_KEY`
- `AWS_DEFAULT_REGION`

```python
import os
import json
```

```python
# Load credentials from environment variables
# Set these before running the notebook:
# os.environ['AWS_ACCESS_KEY_ID'] = 'your-key-here'
# os.environ['AWS_SECRET_ACCESS_KEY'] = 'your-secret-here'
# os.environ['AWS_DEFAULT_REGION'] = 'us-east-2'

required = ['AWS_ACCESS_KEY_ID', 'AWS_SECRET_ACCESS_KEY', 'AWS_DEFAULT_REGION']
for var in required:
    if not os.environ.get(var):
        raise EnvironmentError(f"Missing required environment variable: {var}")
print("Credentials loaded from environment.")
```

```python
from s3_fetch import download_applicant_data, BUCKET_NAME, S3_KEY, LOCAL_OUTPUT

local_path = download_applicant_data(
    bucket=BUCKET_NAME,
    s3_key=S3_KEY,
    local_path=LOCAL_OUTPUT,
)
print(f"File saved to: {local_path}")
```

```python
# Verify the downloaded file
with open(LOCAL_OUTPUT, encoding='utf-8') as fh:
    data = json.load(fh)

print(f"Records loaded: {len(data)}")
print(f"Sample record keys: {list(data[0].keys()) if data else 'empty'}")
```
